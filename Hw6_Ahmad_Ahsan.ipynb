## Task 6. Create a streaming DataFrame to read and output streaming data

```
#6a
# Reading .json files and creating a Streaming DF
streamingDF = spark.readStream \
.schema(dataSchema) \
.option("maxFilesPerTrigger", 1) \
.json("dbfs:/databricks-datasets/definitive-guide/data/activity-data/*.json")
```

```
#6b
# Setting up a streaming query
streamingQuery = streamingDF \
.writeStream \
.queryName('OriginalData') \
.format("memory") \
.outputMode("append") \
.start()
```

```
#6c
from time import sleep
# Viewing the output of the streaming query 5 times
for _ in range(5):
    spark.sql("SELECT * FROM OriginalData").show(5)
    sleep(5)
```

```
#6d
# Stopping the query
streamingQuery.stop()
```

## Task 12. Compute the averages of each column ‘x’, ‘y’, ‘z’ and counts of Device column for each combination of values for ‘Model’ and ‘gt’ columns

```
#12a
from pyspark.sql.functions import avg, count
# Creating a transformation to compute averages and counts
avgDF = streamingDF.groupby('gt','Model').agg(
    avg('x').alias('x_avg'),
    avg('y').alias('y_avg'),
    avg('z').alias('z_avg'),
    count('Device').alias('devicecount')
    )
avgDF
```

```
# Starting the streaming query
gtmodelstatsQuery = avgDF.writeStream \
    .queryName("GTModelStats") \
    .format("memory") \
    .outputMode("complete") \
    .start()
```

```
#12b
# Viewing the streaming output 4 times with a filter on x_avg > 0
for _ in range(4):
    spark.sql("SELECT * FROM GTModelStats WHERE x_avg > 0").show()
    sleep(5)
```

```
#12c
# Stopping the streaming query
gtmodelstatsQuery.stop()
```

## Task 13. Perform a join of a static DataFrame with a streaming DataFrame

```
#13a
# Filtering the historical staticDF as we did for the streamingDF
historicalAvgDF = staticDF.groupby('gt','Model').agg(
    avg('x').alias('x_avg_h'),
    avg('y').alias('y_avg_h'),
    avg('z').alias('z_avg_h')
    )
historicalAvgDF
```

```
#13b
# Computing streaming averages
streamingAvgDF = streamingDF.groupBy("gt", "Model").agg(
    avg("x").alias("x_avg_s"),
    avg("y").alias("y_avg_s"),
    avg("z").alias("z_avg_s")
)
streamingAvgDF
```

```
# Joining historical and streaming averages
joinedDF = streamingAvgDF.join(historicalAvgDF, on=["gt", "Model"])

# Start streaming query
joinedQuery = joinedDF.writeStream \
    .queryName("JoinedData") \
    .format("memory") \
    .outputMode("complete") \
    .start()
```

```
#13c
# Viewing the streaming output 4 times
for _ in range(4):
    spark.sql("SELECT * FROM JoinedData").show()
    sleep(60)
```

```
#13d
# Stopping the streaming query
joinedQuery.stop()
```
